- TF‑IDF (Term Frequency–Inverse Document Frequency) is a numerical representation of text.
- Each document is converted into a vector where each dimension corresponds to a word in the vocabulary, and the value is the TF‑IDF score of that word.

- It balances:
    TF (Term Frequency) → how often a word appears in a document.
    IDF (Inverse Document Frequency) → how rare the word is across all documents.

Applications of TF‑IDF Vectors:
   - Search Engines → Rank documents by relevance to a query.
   - Text Classification → Spam detection, sentiment analysis.
   - Keyword Extraction → Identify important terms in documents.
   - Document Similarity → Compare documents using cosine similarity.
   - Information Retrieval → Core scoring mechanism before embeddings became dominant.

```python
#intermediate example 
from sklearn.feature_extraction.text import TfidfVectorizer

docs = ["I love cats", "I love dogs", "Cats are great pets"]
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)
print(X.toarray())
print(vectorizer.get_feature_names_out())
print("vocalbulary:", vectorizer.vocabulary_)
```

```python
#Advanced Example (Pipeline + Classification)
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(ngram_range=(1,2), stop_words='english')),
    ('model', LogisticRegression())
])

docs = ["Cats are great pets", "Dogs are loyal pets"]
labels = [0, 1]  # 0 = cats, 1 = dogs

pipeline.fit(docs, labels)
print(pipeline.predict(["I love cats"]))  # → 0
```
