What is Bag of Words?
 - Bag of Words (BoW) is a simple text representation technique used in Natural Language Processing (NLP).
 - It treats a document as a “bag” of words, ignoring grammar and word order, but keeping track of word frequency.
 - Each document is represented as a vector of word counts.

 Note: 
 - Count how many times each word in the vocabulary appears in the document.
 - Bag of Words converts text into numerical vectors by counting word frequencies
 - BoW ignores word order → “cats love dogs” and “dogs love cats” look the same.
 - Dimensionality explosion → Large vocabulary = huge sparse vectors.
 - Extensions → TF‑IDF, n‑grams, embeddings (Word2Vec, BERT) improve on BoW.

```python
from sklearn.feature_extraction.text import CountVectorizer
docs =["i love cats", "i love dogs"]
vectorizer  =  CountVectorizer()

matrix = vectorizer.fit_transform(docs)
print(matrix.toarray())
#view the vocabulary
print(vectorizer.get_feature_names_out())
print(vectorizer.vocabulary_)
```

Advanced Example (Pipeline + TF‑IDF)
If you’re using BoW inside a pipeline:

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
docs = ["Cats are great pets", "Dogs are loyal pets"]
pipeline = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('clf', LogisticRegression())   
])
pipeline.fit(docs, [0, 1])  # 0 for cats, 1 for dogs

#access vocabulary from the tf-idf vectorizer
tfidf_vectorizer = pipeline.named_steps['tfidf']
print(tfidf_vectorizer.get_feature_names_out())
```
