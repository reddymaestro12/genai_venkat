```python
word = "welcome"
sentence = "welcome to the world of embeddings"
paragraph = "Embeddings are a way to represent words, sentences, or paragraphs as numerical vectors in a high-dimensional space. They capture semantic meaning and relationships between different pieces of text."
```

```python
#create embedding for the word, sentence, and paragraph
model = SentenceTransformer('all-MiniLM-L6-v2')
for text in [word, sentence, paragraph]:
    embedding = model.encode(text)
    print(f"Embedding for '{text}':\n{embedding}\n")
    print(f"Embedding shape for '{text}': {embedding.shape}\n")
    print(f"Embedding type for '{text}': {type(embedding)}\n")
    print(f"Embedding length for '{text}': {len(embedding)}\n")
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np


class EmbeddingModel:
    def __init__(self, model_name="sentence-transformers/all-MiniLM-L6-v2"):
        self.model = SentenceTransformer(model_name)

    def encode(self, texts, normalize=False):
        """
        Generate embeddings for a string or list of strings.
        """
        return self.model.encode(
            texts,
            convert_to_numpy=True,
            normalize_embeddings=normalize
        )

    def similarity(self, text1, text2):
        """
        Compute cosine similarity between two texts.
        """
        embeddings = self.encode([text1, text2], normalize=True)
        return float(np.dot(embeddings[0], embeddings[1]))

    def batch_similarity(self, texts):
        """
        Compute similarity matrix for a list of texts.
        """
        embeddings = self.encode(texts, normalize=True)
        return np.dot(embeddings, embeddings.T)


# Example usage
if __name__ == "__main__":
    embedder = EmbeddingModel()

    sentences = [
        "The weather is lovely today.",
        "It's a beautiful day.",
        "I like programming."
    ]

    embeddings = embedder.encode(sentences)
    print("Embedding shape:", embeddings.shape)

    similarity = embedder.similarity(
        "The weather is lovely today.",
        "It's a beautiful day."
    )
    print("Similarity:", similarity)

    similarity_matrix = embedder.batch_similarity(sentences)
    print(similarity_matrix)
```
