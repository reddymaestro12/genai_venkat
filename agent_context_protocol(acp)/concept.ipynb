ACP (Agent Context Protocol) is a protocol designed to standardize how AI agents exchange context, memory, state, tasks, and reasoning information during collaboration.

Core Components of ACP: context,memory,state,intermediate results,metadata

Think of it as:

MCP (Model Context Protocol) → LLM-to-tool communication
A2A (Agent2Agent) → Agent ↔ Agent context sharing
ACP (Agent Context Protocol) → Shared Context Between Agents

The main goal is to ensure that when one agent hands off work to another, the receiving agent gets the necessary context rather than starting from scratch.
When to use ACP :
   -  Multi agent RAG systems: each agent shares the context 
   - long runnig workflow : ACP maiantains the shared workflow state
   -  HITL
   - hieararcial agentic systems
